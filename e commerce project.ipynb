```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
file_path = r"D:\Python\e commerce project\Sample - Superstore.csv"
```

```python
df = pd.read_csv(file_path, encoding='latin1')
```

```python
df.head()
```

```python
df.describe()
```

```python
df.info()
```

### Converting date columns

```python
df['Order Date'] = pd.to_datetime(df['Order Date'])
df['Ship Date'] = pd.to_datetime(df['Ship Date'])
```

```python
df.info()
```

```python
df.head()
```

```python
df['Order Month'] = df['Order Date'].dt.month
df['Order Year'] = df['Order Date'].dt.year
df['Order Day of Week'] = df['Order Date'].dt.dayofweek
```

```python
df.head()
```

### Monthly sales analysis

```python
sales_by_month = df.groupby('Order Month')['Sales'].sum().reset_index()
sns.lineplot(x = 'Order Month', y = 'Sales',data = sales_by_month)
plt.title("Monthly Sales Analysis")
plt.show()                
```

### Sales analysis by Category

```python
sales_by_category = df.groupby('Category')['Sales'].sum()
plt.pie(sales_by_category, labels = sales_by_category.index,autopct='%1.1f%%')
plt.title("Sales analysis by Category")
plt.show()
```

### Sales analysis by Sub Category

```python
sales_by_subcategory = df.groupby('Sub-Category')['Sales'].sum().reset_index()
```

```python
sales_by_subcategory
```

```python
plt.figure(figsize=(18,6))
sns.barplot(x = 'Sub-Category', y = 'Sales',data = sales_by_subcategory)
plt.title("Sales Analysis by Sub Category")
plt.show()  
```

### Monthly profit analysis

```python
profit_by_month = df.groupby('Order Month')['Profit'].sum().reset_index()
```

```python
profit_by_month
```

```python
sns.lineplot(x='Order Month',y='Profit',data=profit_by_month)
plt.title('Monthly profit analysis')
plt.show()
```

### Profit by category

```python
profit_by_category = df.groupby('Category')['Profit'].sum().reset_index()
```

```python
profit_by_category
```

```python
profit_by_category = df.groupby('Category')['Profit'].sum()
plt.pie(profit_by_category, labels = profit_by_category.index, autopct='%1.1f%%')
plt.title("Profit by Category")
plt.show()
```

### Profit by sub category

```python
profit_by_subcategory = df.groupby('Sub-Category')['Profit'].sum().reset_index()
```

```python
profit_by_subcategory
```

```python
plt.figure(figsize=(18,6))
sns.barplot(x = 'Sub-Category', y = 'Profit',data = profit_by_subcategory)
plt.title("Profit by Sub Category")
plt.show()
```

### Sales and Profit - Customer Segment

```python
sales_profit_by_segment = df.groupby('Segment').agg({'Sales':'sum','Profit':'sum'}).reset_index() 
```

```python
sales_profit_by_segment
```

```python
plot_data = sales_profit_by_segment.melt(id_vars='Segment',value_vars=['Sales','Profit'],
                                         var_name='Type',value_name='Amount')

plt.figure(figsize=(10,4))

sns.barplot(x='Segment', y='Amount', hue='Type', data=plot_data)
plt.title('Sales and Profit by Customer segment')

plt.show()
```

### sales to profit ratio

```python
sales_profit_by_segment = df.groupby('Segment').agg({'Sales':'sum', 'Profit':'sum'}).reset_index()
sales_profit_by_segment['Sales_to_Profit_Ratio'] = sales_profit_by_segment['Sales']/sales_profit_by_segment['Profit']
print(sales_profit_by_segment[['Segment','Sales_to_Profit_Ratio']])
```
